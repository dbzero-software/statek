```python
import dbzero as db0
```

```python
@db0.memo
class MemoObject:
    value: int = 0

@db0.memo
class ContextStorage:
    local_vars = {}
```

```python
def init_db0():
    db0.init("./dbzero_value")
    db0.open("test-prefix")
    
def execute_test():
    storage = ContextStorage()

    obj = MemoObject()
    storage.local_vars ={"memo_object": obj}
    global_vars = globals()
    code = """memo_object.value = 15
new_local_value=13"""
    exec(code,global_vars,storage.local_vars)
    print(obj.value)
    print(storage.local_vars["new_local_value"])
```

```python
init_db0()
```

```python
execute_test()
```
