```python
import dbzero as db0
from dataclasses import dataclass, field

@db0.memo
class DurableObject:
    value: int = 0

@db0.memo
@dataclass
class Context:
    local_vars: dict = field(default_factory=dict)
```

```python
def init_db0():
    
    db0.open("test-prefix")
    
def execute_test():
    
    db0.init("/my-org/my-app", prefix = "data")
    context_1 = Context({ "local_instance": DurableObject() })
    context_2 = Context({ "local_instance": DurableObject() })

    # Execute code within context_1
    exec("some_instance.value=1132", globals(), context_1.local_vars)
    # Execute code within context_2
    exec("some_instance.value=532;new_variable = 'Agentic memory'", globals(), context_2.local_vars)
    
    print(context_1.local_vars)
    print(context_2.local_vars)    

    """
    Outputs:
    {'local_instance': <__main__.DurableObject object at 0x7f2549711d30>}
    {'local_instance': <__main__.DurableObject object at 0x7f2549711e90>, 'new_variable': 'Agentic memory'}    
    """"
```

```python
init_db0()
```

```python
execute_test()
```
